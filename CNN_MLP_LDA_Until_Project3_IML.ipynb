```python
import os
import numpy as np
from scipy.io import loadmat
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, f1_score
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import seaborn as sns

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

GESTURES = {
    "spher": 0,
    "tip": 1,
    "palm": 2,
    "lat": 3,
    "cyl": 4,
    "hook": 5
}

GESTURE_NAMES = ["Spherical", "Tip", "Palm", "Lateral", "Cylindrical", "Hook"]
GESTURE_COLORS = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8', '#F7DC6F']

def load_single_mat_file(file_path, gestures=GESTURES):
    data = loadmat(file_path)
    participant_data = []

    for gesture, label in gestures.items():
        ch1_key = f"{gesture}_ch1"
        ch2_key = f"{gesture}_ch2"
        if ch1_key not in data or ch2_key not in data:
            print(f"Warning: Missing {gesture} in {os.path.basename(file_path)}")
            continue

        ch1 = np.array(data[ch1_key])
        ch2 = np.array(data[ch2_key])
        if ch1.shape != ch2.shape:
            print(f"Shape mismatch for {gesture} in {os.path.basename(file_path)}")
            continue

        # Each row = one trial (100 per gesture × 5 s ≈ 2500 samples)
        for i in range(ch1.shape[0]):
            n1 = ch1[i] / np.linalg.norm(ch1[i]) if np.linalg.norm(ch1[i]) != 0 else ch1[i]
            n2 = ch2[i] / np.linalg.norm(ch2[i]) if np.linalg.norm(ch2[i]) != 0 else ch2[i]
            participant_data.append({
                "label": label,
                "gesture_name": gesture,
                "channel_1": n1,
                "channel_2": n2
            })
    return participant_data

def load_database2_by_day(base_dir):
    data_per_day = {}
    files = sorted([f for f in os.listdir(base_dir) if f.endswith(".mat")])

    for idx, file in enumerate(files, start=1):
        path = os.path.join(base_dir, file)
        data_day = load_single_mat_file(path)
        data_per_day[f"day{idx}"] = data_day
        print(f" Day {idx}: Loaded {len(data_day)} trials\n")

    return data_per_day

def plot_complete_signals_with_gestures(day_data, day_name="Day 1", max_trials_per_gesture=5):
    """
    Plot complete EMG signals from both channels with gesture labels as shaded regions
    """
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 8), sharex=True)

    # Concatenate signals with gesture information
    ch1_complete = []
    ch2_complete = []
    gesture_segments = []  # (start, end, label, gesture_name)

    current_position = 0

    # Group by gesture and take limited trials
    gesture_counts = {i: 0 for i in range(6)}

    for trial in day_data:
        label = trial["label"]

        # Limit trials per gesture for cleaner visualization
        if gesture_counts[label] >= max_trials_per_gesture:
            continue
        gesture_counts[label] += 1

        ch1 = trial["channel_1"].flatten()
        ch2 = trial["channel_2"].flatten()

        signal_length = len(ch1)
        start_pos = current_position
        end_pos = current_position + signal_length

        # Store segment info
        gesture_segments.append((start_pos, end_pos, label, trial["gesture_name"]))

        # Append signals
        ch1_complete.extend(ch1)
        ch2_complete.extend(ch2)

        current_position = end_pos

    ch1_complete = np.array(ch1_complete)
    ch2_complete = np.array(ch2_complete)
    time_axis = np.arange(len(ch1_complete))

    # Plot Channel 1
    ax1.plot(time_axis, ch1_complete, color='black', linewidth=0.5, alpha=0.7)
    ax1.set_ylabel('Channel 1 Amplitude', fontsize=12)
    ax1.set_title(f'{day_name} - Complete EMG Signals with Gesture Labels', fontsize=14, fontweight='bold')
    ax1.grid(True, alpha=0.3)

    # Plot Channel 2
    ax2.plot(time_axis, ch2_complete, color='black', linewidth=0.5, alpha=0.7)
    ax2.set_ylabel('Channel 2 Amplitude', fontsize=12)
    ax2.set_xlabel('Sample Index', fontsize=12)
    ax2.grid(True, alpha=0.3)

    # Add shaded regions for gestures
    for start, end, label, gesture_name in gesture_segments:
        color = GESTURE_COLORS[label]
        ax1.axvspan(start, end, alpha=0.3, color=color)
        ax2.axvspan(start, end, alpha=0.3, color=color)

    # Create legend
    from matplotlib.patches import Patch
    legend_elements = [Patch(facecolor=GESTURE_COLORS[i], alpha=0.5, label=GESTURE_NAMES[i])
                      for i in range(6)]
    ax1.legend(handles=legend_elements, loc='upper right', ncol=6, fontsize=9)

    plt.tight_layout()
    plt.show()

def sliding_window(data, window_size=128, overlap=0.5):
    """Extract sliding windows from signal"""
    step = int(window_size * (1 - overlap))
    return np.array([data[i:i + window_size] for i in range(0, len(data) - window_size + 1, step)])

def process_day_dataset_for_cnn(day_data, window_size=128, overlap=0.5):
    """
    Process day dataset for CNN input
    Returns: X of shape (n_samples, 2, window_size), y of shape (n_samples,)
    """
    X, y = [], []

    for trial in day_data:
        ch1, ch2 = trial["channel_1"].flatten(), trial["channel_2"].flatten()
        label = trial["label"]

        windows_ch1 = sliding_window(ch1, window_size, overlap)
        windows_ch2 = sliding_window(ch2, window_size, overlap)

        # Stack channels: shape becomes (n_windows, 2, window_size)
        for w1, w2 in zip(windows_ch1, windows_ch2):
            X.append(np.stack([w1, w2], axis=0))  # Stack as (2, window_size)
            y.append(label)

    X, y = np.array(X), np.array(y)
    print(f"Dataset shape: X={X.shape}, y={y.shape}")
    return X, y

# PyTorch Dataset
class EMGDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.FloatTensor(X)
        self.y = torch.LongTensor(y)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

# CNN Model (adjusted for 2 channels instead of 9)
class CNNModel_Modified(nn.Module):
    def __init__(self, n_signals=2):
        super().__init__()
        self.network = nn.Sequential(
            # ----- CNN Layer 1 -----
            nn.Conv1d(in_channels=n_signals, out_channels=64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.MaxPool1d(kernel_size=2),   # L: 128 -> 64
            # ----- CNN Layer 2 -----
            nn.Conv1d(in_channels=64, out_channels=128, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.MaxPool1d(kernel_size=2),   # L: 64 -> 32
            # ----- CNN Layer 3 -----
            nn.Conv1d(in_channels=128, out_channels=256, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.MaxPool1d(kernel_size=2),   # L: 32 -> 16
            nn.Flatten(),
            # Adjusted linear layer
            nn.Linear(256 * 16, 120),
            nn.ReLU(),
            nn.Linear(120, 6)
        )

    def forward(self, xb):
        return self.network(xb)

def train_cnn_model(train_loader, num_epochs=50, verbose=True):
    """Train CNN model"""
    model = CNNModel_Modified(n_signals=2).to(device)
    if verbose:
        print(model)

    # Loss & Optimizer
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)

    # Training tracking
    count = 0
    freq_output_loss = 100
    loss_list = []
    iteration_list = []
    train_accuracy_list = []

    iterator = tqdm(range(num_epochs), desc="Training") if verbose else range(num_epochs)

    for epoch in iterator:
        model.train()
        correct = 0
        total = 0

        for i, (data, labels) in enumerate(train_loader):
            data, labels = data.to(device), labels.to(device)
            optimizer.zero_grad()

            # Forward
            outputs = model(data)

            # Training accuracy
            pred_labels = torch.argmax(outputs, 1)
            total += len(labels)
            correct += (pred_labels == labels).sum()

            loss = criterion(outputs, labels)

            # Backprop + update
            loss.backward()
            optimizer.step()

            count += 1
            if count % freq_output_loss == 0:
                accuracy = correct / float(total)
                loss_list.append(loss.data)
                iteration_list.append(count)
                train_accuracy_list.append(accuracy)

            if verbose and count % 500 == 0:
                accuracy = correct / float(total)
                print(f"Iteration: {count} Loss: {loss.item():.4f} Accuracy: {100 * accuracy.item():.2f}%")

    # Plot training curves
    if verbose and len(iteration_list) > 0:
        iteration_array = np.array(iteration_list)
        loss_array = np.array([l.item() for l in loss_list])
        accuracy_array = np.array([a.item() for a in train_accuracy_list])

        fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
        axes[0].plot(iteration_array, loss_array, linewidth=2, color='#E74C3C')
        axes[0].set_title("Training: Loss vs Iterations", fontsize=14, fontweight='bold')
        axes[0].set_ylabel("Loss", fontsize=12)
        axes[0].set_xlabel("Iterations", fontsize=12)
        axes[0].grid(True, alpha=0.3)

        axes[1].plot(iteration_array, accuracy_array, linewidth=2, color='#27AE60')
        axes[1].set_title("Training: Accuracy vs Iterations", fontsize=14, fontweight='bold')
        axes[1].set_ylabel("Accuracy", fontsize=12)
        axes[1].set_xlabel("Iterations", fontsize=12)
        axes[1].grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

    return model

def evaluate_cnn_model(model, test_loader, return_predictions=True):
    """Evaluate CNN model on test set"""
    model.eval()
    test_pred_labels_list = []
    test_true_labels_list = []

    with torch.no_grad():
        for data, labels in test_loader:
            data, labels = data.to(device), labels.to(device)
            outputs = model(data)
            pred_labels = torch.argmax(outputs, 1)

            test_pred_labels_list.extend(pred_labels.cpu().tolist())
            test_true_labels_list.extend(labels.cpu().tolist())

    test_pred_labels = np.array(test_pred_labels_list)
    test_true_labels = np.array(test_true_labels_list)

    # Metrics
    accuracy = accuracy_score(test_true_labels, test_pred_labels)

    if return_predictions:
        return accuracy, test_pred_labels, test_true_labels
    else:
        return accuracy

def plot_confusion_matrices(confusion_matrices, titles, suptitle="Confusion Matrices"):
    """Plot multiple confusion matrices in a grid"""
    n_matrices = len(confusion_matrices)

    if n_matrices <= 3:
        fig, axes = plt.subplots(1, n_matrices, figsize=(6*n_matrices, 5))
    else:
        rows = (n_matrices + 2) // 3  # Ceiling division
        fig, axes = plt.subplots(rows, 3, figsize=(18, 5*rows))

    if n_matrices == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    for idx, (cm, title) in enumerate(zip(confusion_matrices, titles)):
        ax = axes[idx]
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=GESTURE_NAMES, yticklabels=GESTURE_NAMES,
                   cbar_kws={'label': 'Count'}, ax=ax)
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.set_ylabel('True Label', fontsize=11)
        ax.set_xlabel('Predicted Label', fontsize=11)
        ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

    # Hide extra subplots if any
    for idx in range(n_matrices, len(axes)):
        axes[idx].axis('off')

    fig.suptitle(suptitle, fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

def comprehensive_training_and_evaluation(data_day1, data_day2, data_day3,
                                         window_size=128, overlap=0.5,
                                         batch_size=32, num_epochs=50):
    """
    Comprehensive training and evaluation:
    1. Within-day: Train and test on same day
    2. Cross-day: Train on Day 1, test on all days
    """
    print("\n" + "="*80)
    print("COMPREHENSIVE EMG GESTURE DECODING ANALYSIS")
    print("="*80)

    # Process all days
    print("\n[1/3] Processing Day 1...")
    X1, y1 = process_day_dataset_for_cnn(data_day1, window_size, overlap)
    print("[2/3] Processing Day 2...")
    X2, y2 = process_day_dataset_for_cnn(data_day2, window_size, overlap)
    print("[3/3] Processing Day 3...")
    X3, y3 = process_day_dataset_for_cnn(data_day3, window_size, overlap)

    all_results = {}
    all_cms = []
    all_cm_titles = []

    # Split all days into train/test ONCE
    X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
    X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)
    X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.2, random_state=42)

    # ========== WITHIN-DAY EVALUATION ==========
    print("\n" + "="*80)
    print("PART 1: WITHIN-DAY EVALUATION (Train and Test on Same Day)")
    print("="*80)

    within_day_results = {}
    saved_models = {}  # Store trained models

    train_test_configs = [
        (X1_train, X1_test, y1_train, y1_test, "Day 1"),
        (X2_train, X2_test, y2_train, y2_test, "Day 2"),
        (X3_train, X3_test, y3_train, y3_test, "Day 3")
    ]

    for X_train, X_test, y_train, y_test, day_name in train_test_configs:
        print(f"\n{'='*60}")
        print(f"Training and Testing on {day_name}")
        print('='*60)

        # Create dataloaders
        train_dataset = EMGDataset(X_train, y_train)
        train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
        test_loader = DataLoader(EMGDataset(X_test, y_test), batch_size=batch_size)

        # Train model
        model = train_cnn_model(train_loader, num_epochs=num_epochs, verbose=True)

        # Save the Day 1 model for cross-day analysis
        if day_name == "Day 1":
            saved_models["Day 1"] = model
            print(f"\n✓ Saved {day_name} model for cross-day evaluation")

        # Evaluate
        accuracy, preds, trues = evaluate_cnn_model(model, test_loader)
        cm = confusion_matrix(trues, preds)

        within_day_results[day_name] = accuracy
        all_cms.append(cm)
        all_cm_titles.append(f"{day_name} → {day_name}\nAccuracy: {accuracy*100:.2f}%")

        print(f"\n{day_name} Results:")
        print(f"  Accuracy: {accuracy*100:.2f}%")
        print(f"  F1 Score: {f1_score(trues, preds, average='weighted')*100:.2f}%")

    # ========== CROSS-DAY EVALUATION ==========
    print("\n" + "="*80)
    print("PART 2: CROSS-DAY EVALUATION (Reusing Day 1 Model, Test on All Days)")
    print("="*80)

    # Reuse the Day 1 model that was already trained
    cross_day_model = saved_models["Day 1"]
    print("\n✓ Reusing Day 1 model (already trained in within-day analysis)")
    print("  No additional training needed!\n")

    # Test on TEST sets of all days
    cross_day_results = {}

    test_configs = [
        (X1_test, y1_test, "Day 1 → Day 1 (test)"),
        (X2_test, y2_test, "Day 1 → Day 2 (test)"),
        (X3_test, y3_test, "Day 1 → Day 3 (test)")
    ]

    for X_test, y_test, test_name in test_configs:
        print(f"\n{'='*60}")
        print(f"Evaluating: {test_name}")
        print('='*60)

        test_loader = DataLoader(EMGDataset(X_test, y_test), batch_size=batch_size)
        accuracy, preds, trues = evaluate_cnn_model(cross_day_model, test_loader)
        cm = confusion_matrix(trues, preds)

        cross_day_results[test_name] = accuracy
        all_cms.append(cm)
        all_cm_titles.append(f"{test_name}\nAccuracy: {accuracy*100:.2f}%")

        print(f"\nResults:")
        print(f"  Accuracy: {accuracy*100:.2f}%")
        print(f"  F1 Score: {f1_score(trues, preds, average='weighted')*100:.2f}%")
        print("\nClassification Report:")
        print(classification_report(trues, preds, target_names=GESTURE_NAMES, zero_division=0))

    # ========== PLOT ALL CONFUSION MATRICES ==========
    print("\n" + "="*80)
    print("Plotting All Confusion Matrices...")
    print("="*80)
    plot_confusion_matrices(all_cms, all_cm_titles,
                           suptitle="EMG Gesture Recognition - All Confusion Matrices")

    # ========== PLOT COMPARISON BAR CHARTS ==========
    print("\nPlotting Comparison Charts...")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Within-day results
    days = list(within_day_results.keys())
    accuracies = list(within_day_results.values())
    bars1 = ax1.bar(days, accuracies, color=['#3498DB', '#E74C3C', '#2ECC71'])
    ax1.set_ylabel("Accuracy", fontsize=12)
    ax1.set_title("Within-Day Performance\n(Train & Test on Same Day)", fontsize=14, fontweight='bold')
    ax1.set_ylim(0, 1)
    ax1.grid(True, alpha=0.3, axis='y')
    for bar, acc in zip(bars1, accuracies):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{acc*100:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

    # Cross-day results
    cross_labels = list(cross_day_results.keys())
    cross_accuracies = list(cross_day_results.values())
    bars2 = ax2.bar(cross_labels, cross_accuracies, color=['#2E86AB', '#F6C85F', '#6FB07F'])
    ax2.set_ylabel("Accuracy", fontsize=12)
    ax2.set_title("Cross-Day Performance\n(Train on Day 1, Test on All Days)", fontsize=14, fontweight='bold')
    ax2.set_ylim(0, 1)
    ax2.grid(True, alpha=0.3, axis='y')
    for bar, acc in zip(bars2, cross_accuracies):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{acc*100:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

    plt.tight_layout()
    plt.show()

    # ========== SUMMARY TABLE ==========
    print("\n" + "="*80)
    print("FINAL SUMMARY")
    print("="*80)
    print("\nWithin-Day Results:")
    for day, acc in within_day_results.items():
        print(f"  {day}: {acc*100:.2f}%")

    print("\nCross-Day Results (Trained on Day 1):")
    for test_name, acc in cross_day_results.items():
        print(f"  {test_name}: {acc*100:.2f}%")

    return within_day_results, cross_day_results, cross_day_model


# ========== MAIN EXECUTION ==========
base_location_D2 = '/content/drive/MyDrive/IMLBME_Project_Data/semg_data/Database 2'

# Load data
print("Loading database...")
data_D2_days = load_database2_by_day(base_location_D2)

day1_data = data_D2_days["day1"]
day2_data = data_D2_days["day2"]
day3_data = data_D2_days["day3"]

# Plot complete signals with gesture labels
print("\n" + "="*80)
print("VISUALIZATION: Complete EMG Signals with Gesture Labels")
print("="*80)

print("\nPlotting Day 1 signals...")
plot_complete_signals_with_gestures(day1_data, "Day 1", max_trials_per_gesture=3)

print("\nPlotting Day 2 signals...")
plot_complete_signals_with_gestures(day2_data, "Day 2", max_trials_per_gesture=3)

print("\nPlotting Day 3 signals...")
plot_complete_signals_with_gestures(day3_data, "Day 3", max_trials_per_gesture=3)

# Comprehensive training and evaluation
within_results, cross_results, trained_model = comprehensive_training_and_evaluation(
    day1_data, day2_data, day3_data,
    window_size=128,
    overlap=0.5,
    batch_size=32,
    num_epochs=50
)

print("\n" + "="*80)
print("ANALYSIS COMPLETE!")
print("="*80)
```

# **LDA and MLP**

```python
import os
import numpy as np
from scipy.io import loadmat
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neural_network import MLPClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, f1_score
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

GESTURES = {
    "spher": 0,
    "tip": 1,
    "palm": 2,
    "lat": 3,
    "cyl": 4,
    "hook": 5
}

GESTURE_NAMES = ["Spherical", "Tip", "Palm", "Lateral", "Cylindrical", "Hook"]
GESTURE_COLORS = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8', '#F7DC6F']

def load_single_mat_file(file_path, gestures=GESTURES):
    data = loadmat(file_path)
    participant_data = []

    for gesture, label in gestures.items():
        ch1_key = f"{gesture}_ch1"
        ch2_key = f"{gesture}_ch2"
        if ch1_key not in data or ch2_key not in data:
            print(f"Warning: Missing {gesture} in {os.path.basename(file_path)}")
            continue

        ch1 = np.array(data[ch1_key])
        ch2 = np.array(data[ch2_key])
        if ch1.shape != ch2.shape:
            print(f"Shape mismatch for {gesture} in {os.path.basename(file_path)}")
            continue

        # Each row = one trial (100 per gesture × 5 s ≈ 2500 samples)
        for i in range(ch1.shape[0]):
            n1 = ch1[i] / np.linalg.norm(ch1[i]) if np.linalg.norm(ch1[i]) != 0 else ch1[i]
            n2 = ch2[i] / np.linalg.norm(ch2[i]) if np.linalg.norm(ch2[i]) != 0 else ch2[i]
            participant_data.append({
                "label": label,
                "gesture_name": gesture,
                "channel_1": n1,
                "channel_2": n2
            })
    return participant_data

def load_database2_by_day(base_dir):
    data_per_day = {}
    files = sorted([f for f in os.listdir(base_dir) if f.endswith(".mat")])

    for idx, file in enumerate(files, start=1):
        path = os.path.join(base_dir, file)
        data_day = load_single_mat_file(path)
        data_per_day[f"day{idx}"] = data_day
        print(f" Day {idx}: Loaded {len(data_day)} trials\n")

    return data_per_day

def plot_complete_signals_with_gestures(day_data, day_name="Day 1", max_trials_per_gesture=5):
    """
    Plot complete EMG signals from both channels with gesture labels as shaded regions
    """
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 8), sharex=True)

    # Concatenate signals with gesture information
    ch1_complete = []
    ch2_complete = []
    gesture_segments = []  # (start, end, label, gesture_name)

    current_position = 0

    # Group by gesture and take limited trials
    gesture_counts = {i: 0 for i in range(6)}

    for trial in day_data:
        label = trial["label"]

        # Limit trials per gesture for cleaner visualization
        if gesture_counts[label] >= max_trials_per_gesture:
            continue
        gesture_counts[label] += 1

        ch1 = trial["channel_1"].flatten()
        ch2 = trial["channel_2"].flatten()

        signal_length = len(ch1)
        start_pos = current_position
        end_pos = current_position + signal_length

        # Store segment info
        gesture_segments.append((start_pos, end_pos, label, trial["gesture_name"]))

        # Append signals
        ch1_complete.extend(ch1)
        ch2_complete.extend(ch2)

        current_position = end_pos

    ch1_complete = np.array(ch1_complete)
    ch2_complete = np.array(ch2_complete)
    time_axis = np.arange(len(ch1_complete))

    # Plot Channel 1
    ax1.plot(time_axis, ch1_complete, color='black', linewidth=0.5, alpha=0.7)
    ax1.set_ylabel('Channel 1 Amplitude', fontsize=12)
    ax1.set_title(f'{day_name} - Complete EMG Signals with Gesture Labels', fontsize=14, fontweight='bold')
    ax1.grid(True, alpha=0.3)

    # Plot Channel 2
    ax2.plot(time_axis, ch2_complete, color='black', linewidth=0.5, alpha=0.7)
    ax2.set_ylabel('Channel 2 Amplitude', fontsize=12)
    ax2.set_xlabel('Sample Index', fontsize=12)
    ax2.grid(True, alpha=0.3)

    # Add shaded regions for gestures
    for start, end, label, gesture_name in gesture_segments:
        color = GESTURE_COLORS[label]
        ax1.axvspan(start, end, alpha=0.3, color=color)
        ax2.axvspan(start, end, alpha=0.3, color=color)

    # Create legend
    from matplotlib.patches import Patch
    legend_elements = [Patch(facecolor=GESTURE_COLORS[i], alpha=0.5, label=GESTURE_NAMES[i])
                      for i in range(6)]
    ax1.legend(handles=legend_elements, loc='upper right', ncol=6, fontsize=9)

    plt.tight_layout()
    plt.show()

# ========== FEATURE EXTRACTION FUNCTIONS ==========
def moving_window_rms(signal, window_size=100):
    return np.sqrt(np.convolve(signal ** 2, np.ones(window_size) / window_size, mode='same'))

def rectify_emg(data):
    return np.abs(data)

def rms(x):
    return np.sqrt(np.mean(x ** 2))

def mav(x):
    return np.mean(np.abs(x))

def zc(x, threshold=0.01):
    return np.sum(np.diff(np.sign(x)) != 0)

def ssc(x, threshold=0.01):
    return np.sum(np.diff(np.sign(np.diff(x))) != 0)

def wl(x):
    return np.sum(np.abs(np.diff(x)))

def sliding_window(data, window_size=100, overlap=0.5):
    step = int(window_size * (1 - overlap))
    return np.array([data[i:i + window_size] for i in range(0, len(data) - window_size + 1, step)])

def extract_features(window_ch1, window_ch2):
    feats = []
    for ch in [window_ch1, window_ch2]:
        feats.extend([
            rms(moving_window_rms(rectify_emg(ch))),
            mav(moving_window_rms(rectify_emg(ch))),
            zc(ch),
            ssc(rectify_emg(ch)),
            wl(moving_window_rms(rectify_emg(ch)))
        ])
    return np.array(feats)

def process_day_dataset(day_data, window_size=100, overlap=0.5,
                       scaler=None, pca=None,
                       apply_pca=True, n_components=0.95):
    """
    Process day dataset with feature extraction, z-normalization, and PCA

    Parameters:
    - scaler: If provided, use this scaler (for test data). Otherwise fit new one.
    - pca: If provided, use this PCA (for test data). Otherwise fit new one.
    """
    X, y = [], []
    for trial in day_data:
        ch1, ch2 = trial["channel_1"].flatten(), trial["channel_2"].flatten()
        label = trial["label"]

        windows_ch1 = sliding_window(ch1, window_size, overlap)
        windows_ch2 = sliding_window(ch2, window_size, overlap)

        for w1, w2 in zip(windows_ch1, windows_ch2):
            X.append(extract_features(w1, w2))
            y.append(label)

    X, y = np.array(X), np.array(y)

    # Z-normalization
    if scaler is None:
        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X)
    else:
        X_scaled = scaler.transform(X)

    # PCA dimensionality reduction
    if apply_pca:
        if pca is None:
            pca = PCA(n_components=n_components)
            X_pca = pca.fit_transform(X_scaled)
            print(f"  PCA reduced {X_scaled.shape[1]} → {X_pca.shape[1]} features")
        else:
            X_pca = pca.transform(X_scaled)
        return X_pca, y, scaler, pca
    else:
        return X_scaled, y, scaler, None

def plot_confusion_matrices(confusion_matrices, titles, suptitle="Confusion Matrices"):
    """Plot multiple confusion matrices in a grid"""
    n_matrices = len(confusion_matrices)

    if n_matrices <= 3:
        fig, axes = plt.subplots(1, n_matrices, figsize=(6*n_matrices, 5))
    else:
        rows = (n_matrices + 2) // 3  # Ceiling division
        fig, axes = plt.subplots(rows, 3, figsize=(18, 5*rows))

    if n_matrices == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    for idx, (cm, title) in enumerate(zip(confusion_matrices, titles)):
        ax = axes[idx]
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=GESTURE_NAMES, yticklabels=GESTURE_NAMES,
                   cbar_kws={'label': 'Count'}, ax=ax)
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.set_ylabel('True Label', fontsize=11)
        ax.set_xlabel('Predicted Label', fontsize=11)
        ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

    # Hide extra subplots if any
    for idx in range(n_matrices, len(axes)):
        axes[idx].axis('off')

    fig.suptitle(suptitle, fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

def comprehensive_training_and_evaluation(data_day1, data_day2, data_day3,
                                         window_size=100, overlap=0.5,
                                         classifier_type='MLP'):
    """
    Comprehensive training and evaluation with MLP or LDA
    1. Within-day: Train and test on same day
    2. Cross-day: Train on Day 1, test on all days

    classifier_type: 'MLP' or 'LDA'
    """
    print("\n" + "="*80)
    print(f"COMPREHENSIVE EMG GESTURE DECODING ANALYSIS - {classifier_type}")
    print("="*80)

    all_cms = []
    all_cm_titles = []

    # ========== PROCESS ALL DAYS WITH INDEPENDENT PREPROCESSING ==========
    print("\n[Processing Days with Feature Extraction, Z-norm, and PCA...]")

    # Process each day independently (fit scaler and PCA on each day's training data)
    print("\n[1/3] Processing Day 1...")
    X1_raw, y1_raw = [], []
    for trial in data_day1:
        ch1, ch2 = trial["channel_1"].flatten(), trial["channel_2"].flatten()
        windows_ch1 = sliding_window(ch1, window_size, overlap)
        windows_ch2 = sliding_window(ch2, window_size, overlap)
        for w1, w2 in zip(windows_ch1, windows_ch2):
            X1_raw.append(extract_features(w1, w2))
            y1_raw.append(trial["label"])
    X1_raw, y1_raw = np.array(X1_raw), np.array(y1_raw)

    print("[2/3] Processing Day 2...")
    X2_raw, y2_raw = [], []
    for trial in data_day2:
        ch1, ch2 = trial["channel_1"].flatten(), trial["channel_2"].flatten()
        windows_ch1 = sliding_window(ch1, window_size, overlap)
        windows_ch2 = sliding_window(ch2, window_size, overlap)
        for w1, w2 in zip(windows_ch1, windows_ch2):
            X2_raw.append(extract_features(w1, w2))
            y2_raw.append(trial["label"])
    X2_raw, y2_raw = np.array(X2_raw), np.array(y2_raw)

    print("[3/3] Processing Day 3...")
    X3_raw, y3_raw = [], []
    for trial in data_day3:
        ch1, ch2 = trial["channel_1"].flatten(), trial["channel_2"].flatten()
        windows_ch1 = sliding_window(ch1, window_size, overlap)
        windows_ch2 = sliding_window(ch2, window_size, overlap)
        for w1, w2 in zip(windows_ch1, windows_ch2):
            X3_raw.append(extract_features(w1, w2))
            y3_raw.append(trial["label"])
    X3_raw, y3_raw = np.array(X3_raw), np.array(y3_raw)

    print(f"\nRaw feature shapes: Day1={X1_raw.shape}, Day2={X2_raw.shape}, Day3={X3_raw.shape}")

    # Split all days into train/test
    X1_train_raw, X1_test_raw, y1_train, y1_test = train_test_split(X1_raw, y1_raw, test_size=0.2, random_state=42)
    X2_train_raw, X2_test_raw, y2_train, y2_test = train_test_split(X2_raw, y2_raw, test_size=0.2, random_state=42)
    X3_train_raw, X3_test_raw, y3_train, y3_test = train_test_split(X3_raw, y3_raw, test_size=0.2, random_state=42)

    # ========== WITHIN-DAY EVALUATION ==========
    print("\n" + "="*80)
    print("PART 1: WITHIN-DAY EVALUATION (Train and Test on Same Day)")
    print("="*80)

    within_day_results = {}
    saved_models = {}
    saved_preprocessors = {}

    train_test_configs = [
        (X1_train_raw, X1_test_raw, y1_train, y1_test, "Day 1"),
        (X2_train_raw, X2_test_raw, y2_train, y2_test, "Day 2"),
        (X3_train_raw, X3_test_raw, y3_train, y3_test, "Day 3")
    ]

    for X_train_raw, X_test_raw, y_train, y_test, day_name in train_test_configs:
        print(f"\n{'='*60}")
        print(f"Training and Testing on {day_name}")
        print('='*60)

        # Fit scaler and PCA on training data
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train_raw)

        pca = PCA(n_components=0.95)
        X_train = pca.fit_transform(X_train_scaled)
        print(f"  PCA reduced {X_train_scaled.shape[1]} → {X_train.shape[1]} features")

        # Transform test data with fitted scaler and PCA
        X_test_scaled = scaler.transform(X_test_raw)
        X_test = pca.transform(X_test_scaled)

        # Train classifier
        if classifier_type == 'MLP':
            model = MLPClassifier(hidden_layer_sizes=(100, 50), max_iter=1000, random_state=42)
        elif classifier_type == 'LDA':
            model = LinearDiscriminantAnalysis()
        else:
            raise ValueError("classifier_type must be 'MLP' or 'LDA'")

        model.fit(X_train, y_train)

        # Save Day 1 model and preprocessors
        if day_name == "Day 1":
            saved_models["Day 1"] = model
            saved_preprocessors["scaler"] = scaler
            saved_preprocessors["pca"] = pca
            print(f"  ✓ Saved {day_name} model and preprocessors for cross-day evaluation")

        # Evaluate
        preds = model.predict(X_test)
        accuracy = accuracy_score(y_test, preds)
        cm = confusion_matrix(y_test, preds)

        within_day_results[day_name] = accuracy
        all_cms.append(cm)
        all_cm_titles.append(f"{day_name} → {day_name}\nAccuracy: {accuracy*100:.2f}%")

        print(f"\n  Results:")
        print(f"    Accuracy: {accuracy*100:.2f}%")
        print(f"    F1 Score: {f1_score(y_test, preds, average='weighted')*100:.2f}%")

    # ========== CROSS-DAY EVALUATION ==========
    print("\n" + "="*80)
    print("PART 2: CROSS-DAY EVALUATION (Reusing Day 1 Model, Test on All Days)")
    print("="*80)

    # Reuse Day 1 model and preprocessors
    cross_day_model = saved_models["Day 1"]
    scaler_day1 = saved_preprocessors["scaler"]
    pca_day1 = saved_preprocessors["pca"]

    print("\n✓ Reusing Day 1 model and preprocessors (already trained)")
    print("  No additional training needed!\n")

    cross_day_results = {}

    test_configs = [
        (X1_test_raw, y1_test, "Day 1 → Day 1 (test)"),
        (X2_test_raw, y2_test, "Day 1 → Day 2 (test)"),
        (X3_test_raw, y3_test, "Day 1 → Day 3 (test)")
    ]

    for X_test_raw, y_test, test_name in test_configs:
        print(f"{'='*60}")
        print(f"Evaluating: {test_name}")
        print('='*60)

        # Apply Day 1's preprocessing to test data
        X_test_scaled = scaler_day1.transform(X_test_raw)
        X_test = pca_day1.transform(X_test_scaled)

        # Predict
        preds = cross_day_model.predict(X_test)
        accuracy = accuracy_score(y_test, preds)
        cm = confusion_matrix(y_test, preds)

        cross_day_results[test_name] = accuracy
        all_cms.append(cm)
        all_cm_titles.append(f"{test_name}\nAccuracy: {accuracy*100:.2f}%")

        print(f"\n  Results:")
        print(f"    Accuracy: {accuracy*100:.2f}%")
        print(f"    F1 Score: {f1_score(y_test, preds, average='weighted')*100:.2f}%")
        print("\n  Classification Report:")
        print(classification_report(y_test, preds, target_names=GESTURE_NAMES, zero_division=0))

    # ========== PLOT ALL CONFUSION MATRICES ==========
    print("\n" + "="*80)
    print("Plotting All Confusion Matrices...")
    print("="*80)
    plot_confusion_matrices(all_cms, all_cm_titles,
                           suptitle=f"EMG Gesture Recognition - All Confusion Matrices ({classifier_type})")

    # ========== PLOT COMPARISON BAR CHARTS ==========
    print("\nPlotting Comparison Charts...")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Within-day results
    days = list(within_day_results.keys())
    accuracies = list(within_day_results.values())
    bars1 = ax1.bar(days, accuracies, color=['#3498DB', '#E74C3C', '#2ECC71'])
    ax1.set_ylabel("Accuracy", fontsize=12)
    ax1.set_title(f"Within-Day Performance - {classifier_type}\n(Train & Test on Same Day)",
                  fontsize=14, fontweight='bold')
    ax1.set_ylim(0, 1)
    ax1.grid(True, alpha=0.3, axis='y')
    for bar, acc in zip(bars1, accuracies):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{acc*100:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

    # Cross-day results
    cross_labels = list(cross_day_results.keys())
    cross_accuracies = list(cross_day_results.values())
    bars2 = ax2.bar(cross_labels, cross_accuracies, color=['#2E86AB', '#F6C85F', '#6FB07F'])
    ax2.set_ylabel("Accuracy", fontsize=12)
    ax2.set_title(f"Cross-Day Performance - {classifier_type}\n(Train on Day 1, Test on All Days)",
                  fontsize=14, fontweight='bold')
    ax2.set_ylim(0, 1)
    ax2.grid(True, alpha=0.3, axis='y')
    for bar, acc in zip(bars2, cross_accuracies):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{acc*100:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

    plt.tight_layout()
    plt.show()

    # ========== SUMMARY TABLE ==========
    print("\n" + "="*80)
    print(f"FINAL SUMMARY - {classifier_type}")
    print("="*80)
    print("\nWithin-Day Results:")
    for day, acc in within_day_results.items():
        print(f"  {day}: {acc*100:.2f}%")

    print("\nCross-Day Results (Trained on Day 1):")
    for test_name, acc in cross_day_results.items():
        print(f"  {test_name}: {acc*100:.2f}%")

    return within_day_results, cross_day_results, saved_models["Day 1"]


# ========== MAIN EXECUTION ==========
base_location_D2 = '/content/drive/MyDrive/IMLBME_Project_Data/semg_data/Database 2'

# Load data
print("Loading database...")
data_D2_days = load_database2_by_day(base_location_D2)

day1_data = data_D2_days["day1"]
day2_data = data_D2_days["day2"]
day3_data = data_D2_days["day3"]

# Plot complete signals with gesture labels
print("\n" + "="*80)
print("VISUALIZATION: Complete EMG Signals with Gesture Labels")
print("="*80)

print("\nPlotting Day 1 signals...")
plot_complete_signals_with_gestures(day1_data, "Day 1", max_trials_per_gesture=3)

print("\nPlotting Day 2 signals...")
plot_complete_signals_with_gestures(day2_data, "Day 2", max_trials_per_gesture=3)

print("\nPlotting Day 3 signals...")
plot_complete_signals_with_gestures(day3_data, "Day 3", max_trials_per_gesture=3)

# ========== RUN ANALYSIS WITH MLP ==========
print("\n" + "="*80)
print("RUNNING ANALYSIS WITH MLP CLASSIFIER")
print("="*80)

within_results_mlp, cross_results_mlp, trained_model_mlp = comprehensive_training_and_evaluation(
    day1_data, day2_data, day3_data,
    window_size=100,
    overlap=0.5,
    classifier_type='MLP'
)

# ========== RUN ANALYSIS WITH LDA ==========
print("\n\n" + "="*80)
print("RUNNING ANALYSIS WITH LDA CLASSIFIER")
print("="*80)

within_results_lda, cross_results_lda, trained_model_lda = comprehensive_training_and_evaluation(
    day1_data, day2_data, day3_data,
    window_size=100,
    overlap=0.5,
    classifier_type='LDA'
)

# ========== COMPARISON BETWEEN MLP AND LDA ==========
print("\n" + "="*80)
print("COMPARISON: MLP vs LDA")
print("="*80)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Within-Day Comparison
ax1 = axes[0]
days = ['Day 1', 'Day 2', 'Day 3']
mlp_within = [within_results_mlp[d] for d in days]
lda_within = [within_results_lda[d] for d in days]

x = np.arange(len(days))
width = 0.35

bars1 = ax1.bar(x - width/2, mlp_within, width, label='MLP', color='#3498DB')
bars2 = ax1.bar(x + width/2, lda_within, width, label='LDA', color='#E74C3C')

ax1.set_ylabel('Accuracy', fontsize=12)
ax1.set_title('Within-Day Performance Comparison\n(Train & Test on Same Day)', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(days)
ax1.legend(fontsize=11)
ax1.set_ylim(0, 1)
ax1.grid(True, alpha=0.3, axis='y')

for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{height*100:.1f}%', ha='center', va='bottom', fontsize=9)

# Cross-Day Comparison
ax2 = axes[1]
cross_labels_short = ['Day 1→1', 'Day 1→2', 'Day 1→3']
mlp_cross = list(cross_results_mlp.values())
lda_cross = list(cross_results_lda.values())

x2 = np.arange(len(cross_labels_short))

bars3 = ax2.bar(x2 - width/2, mlp_cross, width, label='MLP', color='#2E86AB')
bars4 = ax2.bar(x2 + width/2, lda_cross, width, label='LDA', color='#F6C85F')

ax2.set_ylabel('Accuracy', fontsize=12)
ax2.set_title('Cross-Day Performance Comparison\n(Train on Day 1, Test on All Days)', fontsize=14, fontweight='bold')
ax2.set_xticks(x2)
ax2.set_xticklabels(cross_labels_short)
ax2.legend(fontsize=11)
ax2.set_ylim(0, 1)
ax2.grid(True, alpha=0.3, axis='y')

for bars in [bars3, bars4]:
    for bar in bars:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{height*100:.1f}%', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("ALL ANALYSES COMPLETE!")
print("="*80)
```
